```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


df = pd.read_csv("D:\Python Courses\Machine-Learning-Projects\Student Graduation Prediction\student_graduation.csv")

from IPython.display import display

display(df.head())
# df.shape
# df.columns
# df.info()
# df.describe()
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()

df['Target'] = le.fit_transform(df['Target'])

sns.heatmap(df.corr(numeric_only=True),annot=False)
```

```python
display(df.head(10))
```

```python
without_std_col = ['Daytime/evening attendance','Displaced','Educational special needs','Debtor','Tuition fees up to date','Gender','Scholarship holder','International']
std_col = df.drop(columns=without_std_col+['Target'])

from sklearn.preprocessing import StandardScaler
scaler = StandardScaler().set_output(transform='pandas') 
scaler_col = scaler.fit_transform(std_col)
# print(scaler_col.columns.tolist)

new_df = np.concatenate((df[without_std_col],scaler_col),axis=1)
```

```python
from sklearn.model_selection import train_test_split

x = new_df
y = df['Target']

x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier()

model.fit(x_train,y_train)
y_pred = model.predict(x_test)

from sklearn.metrics import accuracy_score, recall_score, f1_score,confusion_matrix

print(f"Accuracy Score is: {accuracy_score(y_test,y_pred)}")
print(f"Recall Score is: {recall_score(y_test,y_pred,average='weighted')}")
print(f"F1 Score is: {f1_score(y_test,y_pred,average='weighted')}")
print(f"F1 Score is: {confusion_matrix(y_test,y_pred)}")
```
