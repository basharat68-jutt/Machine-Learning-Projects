```python
import pandas as pd
import numpy as np

df = pd.read_csv('D:\Python Courses\Machine-Learning-Projects\Titanic Servival Prediction\Titanic-Dataset.csv')

print(df.head())
print(df.info())
print(df.describe())
print(df.isnull().sum())
```

```python
print(df.columns)
```

```python
from sklearn.impute import SimpleImputer
imp_age = SimpleImputer()
imp_embarked = SimpleImputer(strategy="most_frequent")

df["Age"] = imp_age.fit_transform(df[["Age"]]).ravel()
df["Embarked"] = imp_embarked.fit_transform(df[["Embarked"]]).ravel()

from sklearn.preprocessing import OneHotEncoder
ohe = OneHotEncoder(sparse_output=False,handle_unknown='ignore').set_output(transform='pandas')

ohe_sex = ohe.fit_transform(df[['Sex']])
ohe_embarked = ohe.fit_transform(df[['Embarked']])

trained_df = df.drop(columns=['PassengerId','Name','Survived','Ticket','Fare','Sex','Embarked','Cabin'])
print(trained_df.columns)
new_df = np.concatenate((trained_df,ohe_sex,ohe_embarked),axis=1)


from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

scaled_df = scaler.fit_transform(new_df)
print(scaled_df)

```
